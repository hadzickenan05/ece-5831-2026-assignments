### Network architecture

| Layer | Nodes | Activation |
|---|---|---|
| Layer 0 (input) | 3 | – |
| Layer 1 (hidden 1) | 5 | sigmoid |
| Layer 2 (hidden 2) | 5 | sigmoid |
| Layer 3 (hidden 3) | 2 | sigmoid |
| Layer 4 (output) | 2 | ReLU |

Lets start off with importing the numpy library and the MultilayerPerceptron class:

```python
import numpy as np
from multilayer_perceptron import MultilayerPerceptron

model = MultilayerPerceptron()
```

Here we are importing numpy library and the class MultilayerPerceptron. I also created a object model which calls the MultilayerPerceptron class to initialize the multilayer perceptron. Next we are going to implement the forward propagation process. The forward propagation process is the process of passing a set of inputs through each layer of a neural network to calculate a final output. At each layer, the input values are multiplied by the weight plus bias of each node in the layer. Then we apply the activation function to produce a output of that layer. The output of one layer will become the input for the next until it reaches the final layer which is the output. Lets create a set of test inputs. Since our architecture requires 3 inputs we will create 4 examples with 3 inputs in each:

```python
inputs = [np.array([1.0, 0.5, -1.0]),
          np.array([0.0, 0.0, 0.0]),
          np.array([2.0, -1.0, 0.5]),
          np.array([-1.5, 2.0, 1.0])]
```

As you can see above we have a list of 4 sets of arrays with 3 inputs for each element. We will test each list of inputs and obtain an output:

```python
for x in inputs:
    y = model.forward(x)
    print("input :", x)
    print("output:", np.round(y, 4))
    print()
```
