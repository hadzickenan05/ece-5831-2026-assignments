## HW #5 - Forward Propagation Implementation: Problem 2
In this problem I implemented a class mnist with a pre-trained neural network (sample_weight.pkl) which was provided by the professor. This neural network is used to classify handwritten digits. The network is a 3 layer network with its inputs being all the pixel values in a 28x28x1 image of a handwritten number. Since the image is 28x28x1 there will be 784 inputs into the neural network. We used the sigmoid acrivation function in the hidden layers and a softwax in the output. 

# Importing Libraries and the Mnist Class
Here we implemented and imported the NumPy and matplotlib.pyplot libraries for calculations and being able to show the image. Also we imported the Mnist class into this code and created a variable network anc called Mnist() to activate the neural network.

```python
import numpy as np
import matplotlib.pyplot as plt
from mnist import Mnist
```

```python
network = Mnist()
```

# Number 0:
We can see that the number 0 was sucessfully detected in all 5 images. 

```python
for m in range(1, 6):
    fname = "0_%d.png" % (m)
    img = network.loadimg(fname)

    plt.imshow(img, cmap='gray')
    plt.title(fname)
    plt.axis('off')
    plt.show()

    y = network.predict(img.reshape(784))
    pred = np.argmax(y)

    if pred == 0:
        print("Success: Image %s is for digit 0 is recognized as %d."% (fname, pred))
        
    else:
        print("Fail: Image %s is for digit 0 but the interface result is %d."% (fname, pred))
```
