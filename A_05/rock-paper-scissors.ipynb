## HW #5 - Forward Propagation Implementation: Problem 1

In this Problem I traned an image classifier with Google's Teachable Machine to recognize rock, paper, and scissors hand gestures for a game of rock-paper-scissors. I exported the trained model in Keras format and loaded it into VS Code. What I did is I resized each of my test images to 224x224. From there I converted each image into an array and scaled it from the range -1 to 1. As I was writing the code, I did some research on the Google Teachable Machine website and found out that I need ot normalize the array from -1 to 1 when importing it into Python or TensorFlow. From here I took the array of data and passed it through the model. The model outputs its prediction and I also made sure to output the image so that there is a visual representation of what is being passed through the model.

# Setup:
Here I set up the code by importing all the necessary libraries needed for the code. I then create the model and I get all the labels needed. I open the generated labels.txt file and read all the classification labels. Since I forgot to rename the files when traing in the Teachable Machine, I renamed it here in the code.

```python
import tensorflow as tf
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

# Load Model and obtain the labels
model = tf.keras.models.load_model("keras_model.h5", compile = False)
labels = open("labels.txt").read().splitlines()

print(labels)

# Rename labels (I forgot to do it in the model training)
for i in range(len(labels)):
    if labels[i] == "0 Class 1":
        labels[i] = "Rock"
    if labels[i] == "1 Class 2":
        labels[i] = "Paper"
    if labels[i] == "2 Class 3":
        labels[i] = "Scissors"

print(labels)
```

# Scissors:
Here I used an image that represents scissors as shown in the output. As mentioned before I first open the image in variable img. From there I resize it and convert the image into an array of numbers. From there I normalize it between -1 and 1 and send it through the model. Before I send it through the model I print the image out. After sending it through the model I get an array of 3 elements each element has a probability float number that tells me the probability of the image classified in that class. There I find the highest value and obtain its index and from there I print out the probable classification. 

```python
img = Image.open("Scissors.png").convert("RGB")
img = img.resize((224, 224))

img_array = np.asarray(img, dtype=np.float32)
img_array = (img_array / 127.5) - 1
img_array = np.expand_dims(img_array, axis = 0)

plt.imshow(img)
plt.axis('off')
plt.show()

#Run through model
predictions = model.predict(img_array)

# Determine the highest probability
predicted_index = np.argmax(predictions[0])
predicted_label = labels[predicted_index]

print("Prediction:", predicted_label)
```

# Paper:

As mentiond before I follow the same exact steps in Scissors to obtain the image that was sent through the model and the predicted classificaiton of the image.

```python
img = Image.open("Paper.png").convert("RGB")
img = img.resize((224, 224))
img_array = np.asarray(img, dtype=np.float32)
img_array = (img_array / 127.5) - 1
img_array = np.expand_dims(img_array, axis = 0)

plt.imshow(img)
plt.axis('off')
plt.show()

predictions = model.predict(img_array)

predicted_index = np.argmax(predictions[0])
predicted_label = labels[predicted_index]

print("Prediction:", predicted_label)

```
